```python
# chatbot -> Application of NLP
# Import libraries

import sklearn
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import re
```

```python
with open("/content/drive/MyDrive/NIAI-AI ML/October/Week 7/Day 25/chatbot_general_qa_40.txt", encoding = "utf-8-sig") as file:
  text = file.read()

print(text[:500])
```

```python
# questions and answers ko extract krna
pattern = r"(?im)^\s*Q:\s*(.*?)\s*\r?\n\s*A:\s*(.*?)\s*$"
#question and answers ki pairs banaye gy (relevent question to relevent answer)
pairs = re.findall(pattern, text)

# all questions aik list me
questions =[q.strip() for q, a in pairs]
# all answers in one list
answers = [a.strip() for q, a in pairs]

# total questions and answers
print("Total Questions and Answers:", len(questions))

for i in range(min(5, len(questions))):
  print("Q:", questions[i])
  print("A:", answers[i])
  print()



```

```python
# Now TFIDF Model Banaingy
from IPython.core.interactiveshell import no_op
# In NLP convert in lowercase and stopwords remove
vectorizer = TfidfVectorizer(lowercase = True, stop_words = "english", ngram_range = (1,2))
# transform meaning pehly khud sekho phir use karo
question_vector = vectorizer.fit_transform(questions)

print("TFIDF is ready to use")
print("questions vector shape:", question_vector.shape)
```

```python
def chatbot(user_question):
  #user k given question ko numbers me convert kna

  user_vector = vectorizer.transform([user_question])

  #user k question or dataset k question ki similarity dekhni

  score = cosine_similarity(user_vector, questions_vector).flatten()

  # best index value nikalni -> hamary chatbot ko similar question find karny me asani ho

  best_index = score.argmax()

  # best matcing question ka best matching answer nikalna hai
  # best score ki base par answer nikalna hai

  best_score = score[best_index]

  if score < 0.2: # threshold
     return "Sorry I don't know the relevent answer"

  return answers[best_index]
```

```python
# Redefining corrected chatbot function to fix the NameError and threshold comparison
def chatbot(user_question):
  user_vector = vectorizer.transform([user_question])
  # Using correct singular variable name 'question_vector'
  score = cosine_similarity(user_vector, question_vector).flatten()
  best_index = score.argmax()
  best_score = score[best_index]

  # Using correct scalar comparison
  if best_score < 0.2:
     return "Sorry I don't know the relevant answer"

  return answers[best_index]

# chatbot ko chalana hai
print("Please ask your question from me relevant to general knowledge")

while True:
  user_question = input("You: ")

  if user_question.lower() in ["exit", "bye", "goodbye", "quit"]:
    print("Good Bye. Have a Nice Day!")
    break

  if not user_question.strip():
    print("Bot: Please Ask Your Next Question")
    continue

  response = chatbot(user_question)
  print("Bot:", response)
```
